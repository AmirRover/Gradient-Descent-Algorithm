## **🔸animate Func**

```python
import matplotlib.animation as animation
from IPython.display import HTML

def animate(x, func, gradfunc, xi, eta, N):
    if gradfunc is None:
        gradfunc = func

    xi_hist = [xi]

    for iteration in range(N):
        xi = GradientDescent(gradfunc, xi, eta, 1)
        xi_hist.append(xi)

    xi_hist = np.array(xi_hist)
    fxi_hist = func(xi_hist)

    fig, ax = plt.subplots(figsize=(8, 6))

    def plot_animate(i):
        ax.clear()

        ax.plot(x, func(x), linewidth=3)
        ax.plot(
            xi_hist[:i+1],
            fxi_hist[:i+1],
            color='red',
            linewidth=4
        )
        ax.plot(
            xi_hist[0],
            fxi_hist[0],
            'r*',
            markersize=15
        )

        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')

    ani = animation.FuncAnimation(
        fig,
        plot_animate,
        frames=len(xi_hist),
        interval=120,
        repeat=False
    )

    plt.close(fig)

    return HTML(ani.to_jshtml())
```

## **🔸Gradient Descent**

### Xi / Learning Rate / Epoch

```python
xi = -3.5
eta = 0.1
epoch = 50
```

```python
def GradientDescent(gradfunc, xi, eta, N):
    for i in range(N):
        grad = gradfunc(xi)
        xi -= eta*grad
    return xi
```

```python
GradientDescent(gradfuncx, xi, eta, epoch)
```
